```python
from IPython.display import HTML, display

nokia_snake_html = """
<div style="display: flex; flex-direction: column; align-items: center; font-family: 'Courier New', monospace; background: #222; padding: 20px; border-radius: 25px; width: 320px; margin: auto; box-shadow: 0 10px 25px rgba(0,0,0,0.5); border: 4px solid #444;">
  <div style="color: #ccc; font-weight: bold; letter-spacing: 3px; margin-bottom: 10px; font-size: 18px;">NOKIA</div>

  <!-- Classic Nokia LCD Screen -->
  <div style="background: #9bbc0f; padding: 10px; border: 6px solid #111; border-radius: 6px; box-shadow: inset 0 0 10px rgba(15,56,15,0.4);">
    <div style="display: flex; justify-content: space-between; color: #0f380f; font-weight: bold; font-size: 14px; border-bottom: 2px solid #0f380f; padding-bottom: 4px; margin-bottom: 6px;">
      <span>SCORE: <span id="score">0000</span></span>
      <span id="status">PLAYING</span>
    </div>
    <canvas id="nokiaScreen" width="240" height="200" style="display: block; background: #9bbc0f;"></canvas>
  </div>

  <!-- On-Screen Keypad Controls -->
  <div style="margin-top: 18px; display: grid; grid-template-columns: repeat(3, 60px); gap: 8px; text-align: center;">
    <div></div>
    <button onclick="setDir('UP')" style="padding: 12px; background: #333; color: white; border: 2px solid #555; border-radius: 8px; cursor: pointer; font-weight: bold;">▲<br><small>2</small></button>
    <div></div>
    <button onclick="setDir('LEFT')" style="padding: 12px; background: #333; color: white; border: 2px solid #555; border-radius: 8px; cursor: pointer; font-weight: bold;">◄<br><small>4</small></button>
    <button onclick="resetGame()" style="padding: 12px; background: #0f380f; color: #9bbc0f; border: 2px solid #555; border-radius: 8px; cursor: pointer; font-size: 11px; font-weight: bold;">RESET</button>
    <button onclick="setDir('RIGHT')" style="padding: 12px; background: #333; color: white; border: 2px solid #555; border-radius: 8px; cursor: pointer; font-weight: bold;">►<br><small>6</small></button>
    <div></div>
    <button onclick="setDir('DOWN')" style="padding: 12px; background: #333; color: white; border: 2px solid #555; border-radius: 8px; cursor: pointer; font-weight: bold;">▼<br><small>8</small></button>
    <div></div>
  </div>
  <div style="color: #888; font-size: 11px; margin-top: 12px;">Click screen & use Arrow Keys / WASD / Buttons</div>
</div>

<script>
  const canvas = document.getElementById("nokiaScreen");
  const ctx = canvas.getContext("2d");
  const box = 10;
  const cols = canvas.width / box;
  const rows = canvas.height / box;

  let snake, food, direction, nextDirection, score, gameLoop, isGameOver;

  function resetGame() {
    snake = [{ x: 10, y: 10 }, { x: 9, y: 10 }, { x: 8, y: 10 }];
    direction = "RIGHT";
    nextDirection = "RIGHT";
    score = 0;
    isGameOver = false;
    document.getElementById("score").innerText = "0000";
    document.getElementById("status").innerText = "PLAYING";
    spawnFood();
    clearInterval(gameLoop);
    gameLoop = setInterval(update, 120);
  }

  function spawnFood() {
    food = {
      x: Math.floor(Math.random() * cols),
      y: Math.floor(Math.random() * rows)
    };
    // Ensure food doesn't spawn on the snake
    if (snake.some(seg => seg.x === food.x && seg.y === food.y)) {
      spawnFood();
    }
  }

  function setDir(newDir) {
    if (newDir === "LEFT" && direction !== "RIGHT") nextDirection = "LEFT";
    if (newDir === "UP" && direction !== "DOWN") nextDirection = "UP";
    if (newDir === "RIGHT" && direction !== "LEFT") nextDirection = "RIGHT";
    if (newDir === "DOWN" && direction !== "UP") nextDirection = "DOWN";
  }

  document.addEventListener("keydown", (e) => {
    if (["ArrowUp", "ArrowDown", "ArrowLeft", "ArrowRight"].includes(e.key)) {
      e.preventDefault(); // Prevent Colab page from scrolling
    }
    if (e.key === "ArrowLeft" || e.key === "a" || e.key === "4") setDir("LEFT");
    if (e.key === "ArrowUp" || e.key === "w" || e.key === "2") setDir("UP");
    if (e.key === "ArrowRight" || e.key === "d" || e.key === "6") setDir("RIGHT");
    if (e.key === "ArrowDown" || e.key === "s" || e.key === "8") setDir("DOWN");
  });

  function update() {
    if (isGameOver) return;
    direction = nextDirection;

    let head = { x: snake[0].x, y: snake[0].y };
    if (direction === "LEFT") head.x--;
    if (direction === "RIGHT") head.x++;
    if (direction === "UP") head.y--;
    if (direction === "DOWN") head.y++;

    // Classic Nokia wall wrap-around (Change to wall collision if preferred)
    if (head.x < 0) head.x = cols - 1;
    if (head.x >= cols) head.x = 0;
    if (head.y < 0) head.y = rows - 1;
    if (head.y >= rows) head.y = 0;

    // Self-collision check
    if (snake.some(seg => seg.x === head.x && seg.y === head.y)) {
      isGameOver = true;
      document.getElementById("status").innerText = "GAME OVER";
      clearInterval(gameLoop);
      return;
    }

    snake.unshift(head);

    // Eat food
    if (head.x === food.x && head.y === food.y) {
      score += 7;
      document.getElementById("score").innerText = String(score).padStart(4, "0");
      spawnFood();
    } else {
      snake.pop();
    }

    draw();
  }

  function draw() {
    // Clear LCD background
    ctx.fillStyle = "#9bbc0f";
    ctx.fillRect(0, 0, canvas.width, canvas.height);

    // Draw pixelated food (classic Nokia cross/diamond food)
    ctx.fillStyle = "#0f380f";
    const fx = food.x * box;
    const fy = food.y * box;
    ctx.fillRect(fx + 3, fy, 4, 10);
    ctx.fillRect(fx, fy + 3, 10, 4);

    // Draw snake segments with slight pixel gap
    snake.forEach((seg, index) => {
      ctx.fillStyle = "#0f380f";
      ctx.fillRect(seg.x * box + 1, seg.y * box + 1, box - 2, box - 2);
    });
  }

  resetGame();
</script>
"""

display(HTML(nokia_snake_html))

```
